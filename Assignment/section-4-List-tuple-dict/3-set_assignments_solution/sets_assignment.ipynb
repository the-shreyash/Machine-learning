```python
### Assignment 3: Set Operations

# Create two sets: one with the first 5 positive integers and another with the first 5 even integers.
#  Perform and print the results of union, intersection, difference, 
# and symmetric difference operations on these sets.
set1 = set()
set2= set()

for i in range(1,6):
    set1.add(i)

print(f"set1: {set1}")
for i in range(10):
    if i%2==0:
        set2.add(i)

print(f"set2: {set2}")

union_s = set1.union(set2)
intersection_s= set1.intersection(set2)
difference_s = set1.difference(set2)
symmertic_diff_s = set1.symmetric_difference(set2)

print(union_s)
print(intersection_s)
print(difference_s)
print(symmertic_diff_s)


```

```python
### Assignment 7: Subsets and Supersets

# Create two sets: one with the first 5 positive integers and another with the first 3 positive integers.
# Check if the second set is a subset of the first set and if the first set is a superset of the second set. 
# Print the results.

set5 = set()
set3 = set()

for i in range(1,6):
    set5.add(i)

for i in range(1,4):
    set3.add(i)

print(set3.issubset(set5))

print(set5.issuperset(set3))


```

```python
### Assignment 9: Set and List Conversion

# Create a set with the first 5 positive integers.
#  Convert it to a list, append the number 6, and convert it back to a set.
#  Print the resulting set.

lst_set = set(range(1,6))
print(type(lst_set))


print(lst_set)
temp_lst = list(lst_set)
temp_lst.append(6)
print(f"temp list {temp_lst}")

lst_set = set(temp_lst)
print(lst_set)
```
