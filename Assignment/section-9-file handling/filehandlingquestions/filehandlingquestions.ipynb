```python
### Assignment 5: Counting Words in a File

# Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

def count_word(filename):
    with open(filename,'r') as file:
        content = file.read()
        count  = len(content.split())

    return count

count_word('sample.txt')
```

```python
### Assignment 6: Finding and Replacing Text

# Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.


def replace_file(filename, old, new):
    with open(filename, 'r+') as file:
        content = file.read()

        new_text = content.replace(old, new)

        file.seek(0)
        file.write(new_text)
        file.truncate()

   


replace_file('sample.txt', 'ceo', 'founder')

```

```python
### Assignment 7: Reading a File in Reverse

# Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

def reverse(filename):
    with open(filename,'r') as file:
        lines = file.readlines()

        for line in lines[::-1]:
            print(line.strip())


reverse('sample.txt')


```

```python
### Assignment 8: Counting Lines, Words, and Characters

# Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words,
#  and characters in the file.

def count_content(filename):
    with open(filename,'r') as file:
        lines = file.readlines()
        line_len = len(lines)
        file.seek(0)
        content = file.read()
        lenght_words = len(content.split())
        
        char_len = len(content)

    return [line_len,lenght_words,char_len]

count_content('sample.txt')

```

```python
### Assignment 11: Creating a Log File

# Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

import datetime as dt

def log_msg(msg, filename = 'activity.log'):
    timestamps = dt.datetime.now().isoformat()

    with open(filename,'a') as file:
        file.write(f'[{timestamps}] {msg}\n')

log_msg('the timestamp')

```

```python
### Assignment 12: Binary File Operations

# Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named
#  `copy_image.bin`.

def bin_transfer(source,desti):
    with open(source,'rb') as src:
        with open(desti,'wb') as des:
            des.write(src.read())
            
```
