```python
##possword strength checker 


def is_strong_password(password):
    """this function check if the password is strong or not """

    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()-+_' for char in password):
        return False
    
    return True


is_strong_password("shreyashS!@12")
```

Calculate the total cost of items in a shopping cart

```python
##possword strength checker 

def cal_total_cost(cart):
    total_cost = 0

    for item in cart:
        total_cost += item['price']*item['quantity']

    return total_cost



#eg of card data

cart = [
    {'name':'apple','price':0.5,'quantity':5},
    {'name':'ipad','price':2,'quantity':2},
    {'name':'apple watch','price':20000,'quantity':1},
    {'name':'macbook','price':150000,'quantity':2},
    {'name':'imac','price':500000,'quantity':3},
    
]

print(type(cart[0]['name']))
print(type(cart))
cal_total_cost(cart)
```

Filter with a lambda function and multiple condtions

```python
numbers = [1,2,3,4,5,6,7,8]
even_and_greater_than_five = list(filter(lambda x:x>5 and x%2==0 ,numbers))
print(even_and_greater_than_five)
```
