```python
class dog:
    ##constructor
    def __init__(self,name,age):
        self.name = name
        self.age = age

##create object

dog1 = dog("buddy",3)
print(dog1)

dog2 = dog("cendy",5)
print(dog2.name)
```

```python
##define a class with instance method(means tell what to do )

class dog:
    def __init__(self,name,age):
        self.name = name
        self.age= age

    def bark(self):
        print(f"{self.name} say wooffff woofff")


#create object

dog1 =dog("buddy",3)
dog1.bark()

dog2 = dog("cendy",5)
dog2.bark()
```

```python
class BankAccount:
    def __init__(self,owner,balance = 0):
        self.owner = owner
        self.balance = balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount } is deposited .now balane is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("insufficient funds")
        else:
            self.balance -+amount
            print(f"{amount } is deposited .now balane is {self.balance}")

##create account

account = BankAccount("shreyas",1000000)
print(account.balance)


##call instance method

account.deposit(1000000)
account.withdraw(10)
```
